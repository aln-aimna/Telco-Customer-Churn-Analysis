# Optional Challenge — NumPy and Customer Tenure

A customer-success manager wants to divide customers into three simple tenure groups:

- **New:** 0–12 months
- **Established:** 13–48 months
- **Long-term:** more than 48 months

Using NumPy:

1. Create a NumPy array of customer tenure.
2. Create Boolean conditions for the three groups.
3. Count the number of customers in each group.
4. For each group, calculate the average monthly charge using the original Pandas data and the corresponding Boolean condition.

**Business question:** Which tenure group represents the largest customer base, and which group has the highest average monthly charge?

```python
# Your work here
tenure = data["tenure"].to_numpy()
```

```python
new = (tenure >= 0) & (tenure <= 12)
established = (tenure >= 13) & (tenure <= 48)
long_term = tenure > 48
```

```python
print("New:", np.sum(new))
print("Established:", np.sum(established))
print("Long-term:", np.sum(long_term))
```

```python
new_avg_charge = data.loc[new, "MonthlyCharges"].mean()
established_avg_charge = data.loc[established, "MonthlyCharges"].mean()
long_term_avg_charge = data.loc[long_term, "MonthlyCharges"].mean()

print("\nAverage Monthly Charge:")
print("New:", new_avg_charge)
print("Established:", established_avg_charge)
print("Long-term:", long_term_avg_charge)
```
